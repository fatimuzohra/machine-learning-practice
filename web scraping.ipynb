```python
rows=table.find_all('tr')
```

```python
print(len(rows))
```

```python
print(rows[0])
```

```python
print(rows[0].get_text(" ", strip=True))
```

```python
headers = rows[0].find_all(["th", "td"])
```

```python
column_names = [h.get_text(" ", strip=True) for h in headers]

print(column_names)
```

```python
data = []

for row in rows[1:]:
    cells = row.find_all(["td", "th"])
    
    row_data = [cell.get_text(" ", strip=True) for cell in cells]
    
    if row_data:
        data.append(row_data)
        
```

```python
print(data[:50])
```
