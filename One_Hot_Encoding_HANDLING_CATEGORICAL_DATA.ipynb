DUMMY VARIABLE TRAP:
The Dummy Variable Trap happens when we use all dummy columns created by One-Hot Encoding, causing redundant informatioN

One-Hot Encoding gives:

| Gender | Male | Female |
| ------ | ---: | -----: |
| Male   |    1 |      0 |
| Female |    0 |      1 |
| Male   |    1 |      0 |
| Female |    0 |      1 |

Notice something:

If we know `Male = 0`, we automatically know:
Female = 1
And if:
Male = 1
Female = 0
So we don't actually need **both columns**.

| fuel   | fuel_Diesel | fuel_Petrol |

### How do we know CNG?

If:

fuel_Diesel = 0
fuel_Petrol = 0
then it must be:

CNG

So we don't need a separate `fuel_CNG` column.

4. ONE-HOTENCODING WITH TOP CATEGORIES

````python
counts = df['brand'].value_counts()
#Count how many times each brand appears
````

````python
df['brand'].nunique()
threshold = 100
#nunique() = number of unique values
#If a brand appears 100 times or less, we will consider it uncommon.
````

````python
#Find the uncommon brands
#repl is basically a list of uncommon categories.
repl = counts[counts <= threshold].index
````

````python
pd.get_dummies(df['brand'].replace(repl, 'uncommon')).sample(5)
````

df['brand'].replace(repl, 'uncommon')
