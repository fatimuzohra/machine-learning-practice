
# Day 01 — Introduction to Machine Learning

## 1. What is Machine Learning?

**Machine Learning (ML)** is a branch of Artificial Intelligence (AI) that allows computers to **learn patterns from data and make predictions or decisions without being explicitly programmed for every situation**.

### Simple example

Suppose we want to predict whether an email is **Spam** or **Not Spam**.

We could manually write rules such as:

If email contains "WIN MONEY" → Spam
If email contains "FREE PRIZE" → Spam
If email contains "CONGRATULATIONS" → Spam
```

But there can be thousands of different ways a spam email can be written.

Instead of creating every rule manually, we can give a Machine Learning algorithm:

Past emails + their labels
            ↓
     Machine Learning
            ↓
     Learns patterns
            ↓
   New email → Spam / Not Spam
```

So, the machine **learns from examples**.

# 2. Why Do We Need Machine Learning?

Traditional programming works very well when we can clearly define the rules.

But many real-world problems are too complicated to solve by writing rules manually.

### Example: House Price Prediction

Suppose we want to predict the price of a house.

The price may depend on:

* Area
* Number of bedrooms
* Location
* Number of bathrooms
* Age of the house
* Parking
* Nearby facilities
* etc.

Writing a perfect rule like this would be extremely difficult:

Price = Rule 1 + Rule 2 + Rule 3 + ...
```

Instead, we can provide historical house data:

|       Area | Bedrooms | Location | Price |
| ---------: | -------: | -------- | ----: |
| 1000 sq ft |        2 | City     | ₹50 L |
| 1500 sq ft |        3 | City     | ₹75 L |
| 2000 sq ft |        4 | City     | ₹1 Cr |

The Machine Learning algorithm can learn the relationship between the **features** and the **price**.

Then we can give it a new house:

Area = 1800 sq ft
Bedrooms = 3
Location = City
```

and the model can predict:

Predicted Price = ₹85 L
```

### Machine Learning is useful when:

* There is a large amount of data.
* The patterns are difficult to define manually.
* The problem changes over time.
* We want computers to learn from previous examples.
* We need predictions or automated decisions.

---

# 3. Traditional Programming vs Machine Learning

This is one of the **most important concepts to understand**.

## Traditional Programming

In traditional programming:

Input + Rules
     ↓
  Program
     ↓
  Output
```

For example:

```text
Input: Marks
Rule: If marks >= 40 → Pass
      If marks < 40 → Fail
Output: Pass / Fail
```

We explicitly provide the rules.

---

## Machine Learning

                In Machine Learning:
                
                ```text
                Input + Output
                      ↓
                Machine Learning Algorithm
                      ↓
                     Model
                      ↓
                New Input
                      ↓
                Prediction
```

For example:

                ```text
                Past house data
                     ↓
                ML Algorithm
                     ↓
                Trained Model
                     ↓
                New house
                     ↓
                Predicted price
                ```

The algorithm **learns the relationship from the data** instead of us manually writing every rule.

### Easy comparison

| Traditional Programming        | Machine Learning                          |
| ------------------------------ | ----------------------------------------- |
| Rules are explicitly written   | Patterns are learned from data            |
| Programmer defines the logic   | Algorithm learns the logic                |
| Input + Rules → Output         | Input + Output → Model                    |
| Good for well-defined problems | Useful for complex pattern-based problems |

### Remember

**Traditional Programming:**

> We give the computer the rules.

**Machine Learning:**

> We give the computer examples, and it learns the patterns.

---

# 4. How Machine Learning Works

The basic Machine Learning process can be understood in a few steps.
                    
                    ```text
                    1. Collect Data
                           ↓
                    2. Clean & Prepare Data
                           ↓
                    3. Select Features
                           ↓
                    4. Train Model
                           ↓
                    5. Test Model
                           ↓
                    6. Evaluate Model
                           ↓
                    7. Make Predictions
                    ```

Let's understand each step.

### Step 1 — Collect Data

First, we need data.

Example:

House Dataset
```

containing:

* Area
* Bedrooms
* Location
* Price

---

### Step 2 — Clean & Prepare Data

Real-world data is usually not perfect.

It may contain:

* Missing values
* Duplicate values
* Incorrect data
* Outliers
* Different formats

We prepare the data before giving it to the model.

---

### Step 3 — Select Features

**Features** are the input variables used by the model.

For house price prediction:

Area
Bedrooms
Bathrooms
Location
```

are examples of features.

---

### Step 4 — Train the Model

We give historical data to a Machine Learning algorithm.

The algorithm looks for patterns.
    
    ```text
    Training Data
          ↓
    ML Algorithm
          ↓
    Trained Model
    ```

---

### Step 5 — Test the Model

After training, we give the model data that it has not seen before.

This helps us determine whether the model can **generalize** to new data.

---

### Step 6 — Evaluate the Model

We measure how well the model performs.

Different problems use different metrics.

For example:

* Regression → MAE, MSE, RMSE, R²
* Classification → Accuracy, Precision, Recall, F1 Score

---

### Step 7 — Make Predictions

Once the model performs well enough, we can use it on new data.
        
         New Data
           ↓
        Trained Model
           ↓
        Prediction
        ```

---

# 5. Types of Machine Learning

Machine Learning is commonly divided into three major types:

                         Machine Learning
                               |
                  ┌────────────┼────────────┐
                  ↓            ↓            ↓
             Supervised    Unsupervised  Reinforcement
              Learning       Learning       Learning
        ```

# 5.1 Supervised Learning

In **Supervised Learning**, the model learns from data where the **correct answer is already known**.

In other words, the training data contains:

Input → Correct Output


### Example

Suppose we have:

| Hours Studied | Result |
| ------------: | ------ |
|             2 | Fail   |
|             4 | Pass   |
|             6 | Pass   |
|             8 | Pass   |

The model learns the relationship between:

Hours Studied → Result

Then we give:

Hours Studied = 5


The model might predict:

Pass


### Two major types of Supervised Learning

#### 1. Regression

Used when the output is a **continuous numerical value**.

Examples:

* House price prediction
* Temperature prediction
* Salary prediction
* Stock price prediction

Example:

Input → House details
Output → ₹75,00,000
```

---

#### 2. Classification

Used when the output belongs to a **category/class**.

Examples:

* Spam / Not Spam
* Pass / Fail
* Disease / No Disease
* Cat / Dog

Example:

Input → Email
Output → Spam
```

### Remember

> **Supervised Learning = Data has known answers (labels).**

---

# 5.2 Unsupervised Learning

In **Unsupervised Learning**, the data does **not have predefined answers or labels**.

The model tries to discover patterns or structures in the data by itself.

Example:

Suppose a shopping website has customer information:

| Customer | Age | Spending |
| -------- | --: | -------: |
| A        |  20 |      Low |
| B        |  22 |      Low |
| C        |  45 |     High |
| D        |  48 |     High |

We don't tell the model which customers belong together.

The algorithm may discover:

Group 1 → Younger, low spending
Group 2 → Older, high spending
```

This process is called **clustering**.

### Common Unsupervised Learning techniques

* K-Means Clustering
* Hierarchical Clustering
* DBSCAN
* PCA

### Remember

> **Unsupervised Learning = No predefined answers; the model finds patterns in the data.**

---

# 5.3 Reinforcement Learning

**Reinforcement Learning (RL)** is a type of Machine Learning where an **agent learns by interacting with an environment**.

The agent takes an action and receives:

* **Reward** → good action
* **Penalty** → bad action

The goal is to learn a strategy that maximizes the total reward.

### Simple example

Imagine training a robot to reach a destination.
            
            ```text
            Robot
              ↓
            Takes action
              ↓
            Environment
              ↓
            Reward / Penalty
              ↓
            Robot learns
            ```

If the robot moves toward the destination:

+10 reward
```

If it moves in the wrong direction:

-5 penalty
```

Over time, the robot learns which actions are better.

### Applications

* Game playing
* Robotics
* Autonomous systems
* Recommendation systems
* Resource optimization

### Remember

> **Reinforcement Learning = Learning through actions, rewards, and penalties.**

# 7. Important Machine Learning Terminology

These terms are **very important** because you will see them throughout your ML journey.

## Dataset

A **dataset** is a collection of data used for analysis or Machine Learning.

Example:

Titanic Dataset
House Price Dataset
Customer Dataset
```

---

## Feature

A **feature** is an input variable used by the model.

Example:

For house price prediction:

Area
Bedrooms
Bathrooms
Location


are features.



## Label / Target

The **target** or **label** is the value we want the model to predict.

Example:

```text
Features → Area, Bedrooms, Location
Target   → House Price
```

---

## Sample / Instance

A **sample** is one individual observation or row in a dataset.

Example:

| Age |  Salary | Purchased |
| --: | ------: | --------- |
|  25 | ₹30,000 | Yes       |

This complete row is one sample.

---

## Model

A **model** is the learned representation of patterns from the training data.

Data
 ↓
Algorithm
 ↓
Model


---

## Training

**Training** is the process of teaching the model using data.

```text
Training Data → Model
```

---

## Testing

**Testing** means checking the model using data that it has not seen during training.

```text
Unseen Data → Model → Prediction
```

---

## Prediction

The output produced by a trained Machine Learning model is called a **prediction**.

Example:

```text
Input → House details
Model → Prediction
Output → ₹80 lakh
```

---

## Algorithm

An **algorithm** is the method used to learn patterns from data.

Examples:

* Linear Regression
* Logistic Regression
* Decision Tree
* Random Forest
* KNN

---

## Training Data

Data used to train the Machine Learning model.

---

## Testing Data

Data used to evaluate how well the trained model performs on unseen examples.
