# 🔹 What is Categorical Data?

In a dataset, some columns contain **categories/words instead of numbers**.

For example:

* `Name` → categorical
* `Gender` → categorical
* `City` → categorical
* `Purchased` → categorical

So we need to convert:

**Words → Numbers**

This process is called **Encoding**.


# 🔥 Main Types of Encoding

You should mainly know these:

1. **Label Encoding**
2. **Ordinal Encoding**
3. **One-Hot Encoding**
4. **Dummy Encoding** — basically one-hot encoding with one column removed
5. **Target Encoding** — more advanced

For beginner ML, focus heavily on **Label Encoding and One-Hot Encoding**.

---

# 1️⃣ Label Encoding

Label Encoding converts each category into a number.

Example:
Male
Female
Male
Female
```

After encoding:

```text
Male   → 0
Female → 1
```

# 2️⃣ Ordinal Encoding

Ordinal encoding is useful when categories have a **natural order**.

For example:
School
Bachelor
Master
PhD
```

There is an order:

```text
School < Bachelor < Master < PhD
```

So we can encode:

```text
School   → 0
Bachelor → 1
Master   → 2
PhD      → 3
```

Here, numbers actually represent the order.

# 3️⃣ One-Hot Encoding ⭐⭐⭐
Suppose we have a column called **Color**:

| Color |
| ----- |
| Red   |
| Blue  |
| Green |
| Red   |

A Machine Learning model cannot directly understand `"Red"`, `"Blue"`, `"Green"`.

So we convert them into **0 and 1**.

### Step 1: Create separate columns

Red
Blue
Green
```

### Step 2: Put 1 for the matching color, otherwise 0

| Color | Red | Blue | Green |
| ----- | --: | ---: | ----: |
| Red   |   1 |    0 |     0 |
| Blue  |   0 |    1 |     0 |
| Green |   0 |    0 |     1 |
| Red   |   1 |    0 |     0 |

That's **One-Hot Encoding**. 🎯

# 4️⃣ Dummy Encoding
Let's use the same example:

| Color |
| ----- |
| Red   |
| Blue  |
| Green |
| Red   |

### One-Hot Encoding

We would create **3 columns**:

| Color | Red | Blue | Green |
| ----- | --: | ---: | ----: |
| Red   |   1 |    0 |     0 |
| Blue  |   0 |    1 |     0 |
| Green |   0 |    0 |     1 |
| Red   |   1 |    0 |     0 |

---

## Dummy Encoding

In **dummy encoding, we remove one column**.

For example, remove `Green`:

| Color | Red | Blue |
| ----- | --: | ---: |
| Red   |   1 |    0 |
| Blue  |   0 |    1 |
| Green |   0 |    0 |
| Red   |   1 |    0 |

### 🧠 Why does Green become `0, 0`?

Because:
Red = 0
Blue = 0
```

So we know it must be:
Green
```

Therefore, we don't need a separate Green column.

# 5️⃣ Target Encoding

> **Replace each category with the average value of the target for that category.**

### Example

Suppose we want to predict **whether a student passes**.

| City   | Passed |
| ------ | -----: |
| Delhi  |      1 |
| Delhi  |      0 |
| Mumbai |      1 |
| Mumbai |      1 |
| Delhi  |      1 |

Here:

* **City** = categorical feature
* **Passed** = target

---

### Step 1: Find the average target for each city

#### Delhi

Delhi students:

```text
1, 0, 1
```

Average:

```text
(1 + 0 + 1) / 3 = 0.67
```

So:

**Delhi → 0.67**

#### Mumbai

Mumbai students:

```text
1, 1
```

Average:

```text
(1 + 1) / 2 = 1.00
```

So:

**Mumbai → 1.00**

---

### Step 2: Replace the city with its average

Original:

| City   | Passed |
| ------ | -----: |
| Delhi  |      1 |
| Delhi  |      0 |
| Mumbai |      1 |
| Mumbai |      1 |
| Delhi  |      1 |

After target encoding:

| City   | Encoded City | Passed |
| ------ | -----------: | -----: |
| Delhi  |         0.67 |      1 |
| Delhi  |         0.67 |      0 |
| Mumbai |         1.00 |      1 |
| Mumbai |         1.00 |      1 |
| Delhi  |         0.67 |      1 |

So:

```text
Delhi  → 0.67
Mumbai → 1.00
```

### 🧠 Super easy way to remember

**Target Encoding = Category → Average Target**

```text
Delhi  → average Pass value → 0.67
Mumbai → average Pass value → 1.00
```

### ⚠️ Important

Target encoding needs care because the target values are being used to create the feature. If done incorrectly, it can cause **data leakage**. In real ML projects, calculate the encodings using the training data appropriately rather than using the entire dataset before splitting.

**One-line definition:**

> Target encoding replaces each categorical value with a statistic (usually the mean) of the target variable for that category.



````python
import numpy as np
import pandas as pd
````

````python
df=pd.read_csv('customer.csv')
````

````python
df.sample(5)
````

````python
df=df.iloc[:,2:]
````

````python
df.head()
````

````python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
# Take all rows from the last column.->[:,-1]
````

````python
from sklearn.preprocessing import OrdinalEncoder
#Ordinal Encoding converts categories into numbers when the categories have a meaningful order.
````

````python
x_train
````

````python
oe=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
#we have given poor less value then good
````

````python
oe.fit(x_train)
````

````python
x_train=oe.transform(x_train)
x_test=oe.transform(x_test)
````

| Encoder            | Common use                        | Example                                     |
| ------------------ | --------------------------------- | ------------------------------------------- |

| **OrdinalEncoder** | Feature `X` with meaningful order | Poor → 0, Average → 1, Good → 2             |
| **OneHotEncoder**  | Feature `X` with no order         | Delhi, Mumbai, Bangalore → separate columns |
